```python
import os
from pathlib import Path

from dotenv import load_dotenv
from langchain_cohere import ChatCohere
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.prompts import PromptTemplate


# Load .env from this project folder
env_path = Path.cwd() / ".env"
load_dotenv(env_path)

COHERE_API_KEY = os.getenv("COHERE_API_KEY")

if not COHERE_API_KEY:
    raise ValueError("COHERE_API_KEY is missing from .env")

print("Cohere API key loaded successfully.")
```

```python
speech = """
...
"""
```

```python
print(len(speech))
```

```python
llm = ChatCohere(
    model="command-a-03-2025",
    temperature=0.3,
    cohere_api_key=COHERE_API_KEY
)

print("Cohere summarization model loaded successfully.")
```

```python
messages = [
    SystemMessage(
        content=(
            "You are an expert assistant specializing in "
            "summarizing speeches. "
            "Create a short, clear and accurate summary. "
            "Do not add information that is not present in the speech."
        )
    ),
    HumanMessage(
        content=f"""
Please provide a short and concise summary of the following speech.

TEXT:

{speech}
"""
    )
]

response = llm.invoke(messages)

print(response.content)
```

```python
generic_template = """
Write a precise and concise summary of the following speech.

Speech:
{speech}

Translate the summary into {language}.

Do not add information that is not present in the speech.
"""

prompt = PromptTemplate(
    input_variables=["speech", "language"],
    template=generic_template
)

complete_prompt = prompt.format(
    speech=speech,
    language="Hindi"
)

response = llm.invoke(complete_prompt)

print(response.content)
```
